```python
#Write a python script that takes the radius of a circle as input and displays the diameter, circumference and area of the circle.
import math
r=float(input("Enter radius of circle:"))
d=2*r
p=2*math.pi*r
a=math.pi*(r**2)
print(f"For a circle of radius {r}:\nDiam={d}\nCircumference={p}\narea={a}")
```

```python
"""Create a program that reads a letter of the alphabet from the user. If the user enters a, e, i, o or u
then your program should display a message indicating that the entered letter is a vowel. If the user
enters y then your program should display a message indicating that sometimes y is a vowel, and
sometimes y is a consonant. Otherwise your program should display a message indicating that the letter
is a consonant."""

alph=input("Enter an english alphabet:")
alph=alph.lower()
print(alph)
A='a'
E='e'
O='o'
I='i'
U='u'
Y='y'
if(alph == A or alph==E or alph == O or alph == I or alph == U):
   print(f"{alph} is a vowel")
elif(alph == Y):
   print(f"sometimes y is a vowel and sometimes y is a consonant")
else:
  print(f"{alph} is a consonant")

```

```python
'''Write a program that determines the name of a shape from its number of sides. Read the number of
sides from the user and then report the appropriate name as part of a meaningful message. Your
program should support shapes with anywhere from 3 up to (and including) 10 sides. If a number of
sides outside of this range is entered then your program should display an appropriate error message.'''
num=int(input("Enter no. of sides of shape (3<=n<=10):"))
list=["triangle","square or rhombus or rectangle or parallelogram and so on","pentagon","hexagon","heptagon","octagon","nonagon","decagon"]
if(num < 3 or num > 10):
  print(f"error:the shape with no. of sides = {num} is beyond the limit of this script")
else:
  p=list[num-3]
  print(f"{p}")
```

```python
'''Create a program that reads the name of a month from the user as a string. Then your program should
display the number of days in that month. Display “28 or 29 days” for February .'''
month=input("Enter a month(eg:February):")
month=month.lower()
if(month == 'january' or month == 'march' or month == 'may' or month == 'july' or month =='august' or month == 'october' or month == 'december'):
  print(f"{month} has 31 days")
elif(month == 'february'):
  print(f"{month} has 28 or 29 days")
else:
  print(f"{month} has 30 days")
```
